# Lab 01 — Math + NumPy
Verify dot products, matrix shapes, norms, and a hand-computed reference.


```python
import numpy as np
x = np.array([1.0, 2.0, -1.0])
w = np.array([0.5, -2.0, 3.0])
dot = x @ w
norm = np.linalg.norm(x)
print("dot =", dot)
print("norm =", norm)
assert np.isclose(dot, 1*0.5 + 2*(-2) + (-1)*3)
```
