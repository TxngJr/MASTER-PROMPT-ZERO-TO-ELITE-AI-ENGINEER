# Lab 08 — Final LLM Capstone
Create causal input/target windows and verify the one-token shift.


```python
tokens = list(range(12))
context = 4
pairs = []
for i in range(len(tokens)-context):
    x = tokens[i:i+context]
    y = tokens[i+1:i+context+1]
    pairs.append((x,y))
print(pairs[:3])
for x,y in pairs:
    assert x[1:] == y[:-1]
```
