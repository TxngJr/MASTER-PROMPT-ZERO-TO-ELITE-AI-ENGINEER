# Lab 03 — Backprop
Compare an analytic derivative with a central finite difference.


```python
def f(x):
    return x**3 + 2*x
x = 1.7
eps = 1e-5
numeric = (f(x+eps)-f(x-eps))/(2*eps)
analytic = 3*x*x + 2
print(numeric, analytic, abs(numeric-analytic))
assert abs(numeric-analytic) < 1e-7
```
