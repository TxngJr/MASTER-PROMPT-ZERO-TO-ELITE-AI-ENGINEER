# Lab 02 — Classical ML
Fit a one-feature linear regression with explicit gradient descent.


```python
import numpy as np
rng = np.random.default_rng(7)
x = np.linspace(-1, 1, 32)
y = 2.5 * x - 0.3
w, b, lr = 0.0, 0.0, 0.1
for step in range(200):
    pred = w*x + b
    err = pred-y
    w -= lr * (2/len(x)) * np.sum(err*x)
    b -= lr * (2/len(x)) * np.sum(err)
print(w, b, np.mean((w*x+b-y)**2))
assert abs(w-2.5) < 1e-2 and abs(b+0.3) < 1e-2
```
