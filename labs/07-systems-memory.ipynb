# Lab 07 — Systems Memory
Estimate weight and KV-cache bytes instead of guessing whether a model fits.


```python
def weight_bytes(params, bytes_per_param):
    return params * bytes_per_param

def kv_bytes(layers, batch, seq, kv_heads, head_dim, bytes_per_value):
    return 2 * layers * batch * seq * kv_heads * head_dim * bytes_per_value

print("100M FP16 weights GB:", weight_bytes(100_000_000, 2)/1e9)
print("KV MiB:", kv_bytes(12,1,2048,4,64,2)/1024**2)
assert weight_bytes(100,2) == 200
```
