# Lab 06 — LLM Post-Training
See how an assistant-only mask changes which token losses contribute.


```python
import numpy as np
token_losses = np.array([0.2, 0.4, 1.1, 0.7, 0.5])
assistant_mask = np.array([0, 0, 1, 1, 1], dtype=bool)
masked_mean = token_losses[assistant_mask].mean()
full_mean = token_losses.mean()
print("assistant-only:", masked_mean, "full:", full_mean)
assert not np.isclose(masked_mean, full_mean)
```
