# Lab 05 — RAG + Bounded Tools
Build a tiny lexical retriever and validate an allowlisted tool request.


```python
docs = [
    "gradient descent updates parameters using gradients",
    "RAG retrieves external context before generation",
    "a tokenizer converts text into token identifiers",
]
query = set("how does rag retrieve context".lower().split())
scores = [len(query & set(d.lower().split())) for d in docs]
best = max(range(len(docs)), key=scores.__getitem__)
print("retrieved:", docs[best])
allowed_tools = {"lookup"}
request = {"tool":"lookup","query":"RAG"}
assert request["tool"] in allowed_tools
assert isinstance(request["query"], str) and len(request["query"]) <= 100
```
