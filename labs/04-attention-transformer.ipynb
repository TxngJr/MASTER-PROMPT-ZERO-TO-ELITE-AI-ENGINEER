# Lab 04 — Attention
Compute a tiny scaled dot-product attention example and inspect row normalization.


```python
import numpy as np
Q = np.array([[1.,0.],[0.,1.]])
K = np.array([[1.,0.],[0.,1.]])
V = np.array([[2.,0.],[0.,4.]])
scores = Q @ K.T / np.sqrt(Q.shape[-1])
scores -= scores.max(axis=-1, keepdims=True)
weights = np.exp(scores)
weights /= weights.sum(axis=-1, keepdims=True)
out = weights @ V
print(weights)
print(out)
assert np.allclose(weights.sum(axis=-1), 1.0)
```
